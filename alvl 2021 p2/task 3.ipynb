```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None
        
        
    #getset
    def getdata(self):
        return self.data
    
    def setdata(self, new):
        self.data = new
        return True
    
    def getnext(self):
        return self.next
    def setnext(self, new):
        self.next = new
        
        
        

class LinkedList:
    def __init__(self):
        self.head = None
        
        
    def is_empty(self):
        return self.head == None
        
    def insert(self, integer_value):
        newnode = Node(integer_value)
        
        if self.is_empty():
            self.head = newnode
            
            
        else:
            
           
            
            newnode.next = self.head
            
            self.head = newnode
            
            
    def delete(self, integer_value):
        if self.is_empty():
            return False
        
        current = self.head
        prev = None
        while current != None and current.data != integer_value:
            prev = current
            current = current.next
            
            
        if prev == None: #at the first
         
            self.head = self.head.next
            return True
            
            
        if current.data == integer_value:
            prev.next = current.next
            return True
        
        if current == None:
            return None
        
        
        
        
        
    def search(self, integer_value):
        
        if self.is_empty():
            return False
        
        
        current = self.head 
        prev = None
        
        while current != None and current.data != integer_value:
            prev = current
            current = current.next
            
        if prev == None:
            return True
           
        if current == None:
            return False
        
        if current.data == integer_value:
            return True
        
     
        
        
        
    def count(self):
        if self.is_empty():
            return 0
        
        
        current  = self.head
        prev = None
        count =0
        
        while current != None:
            count += 1
            prev = current
            current = current.next
            
            
            
        return count
    
    
    def to_String(self):
        List = []
        
        current  = self.head
        prev = None
    
        
        while current != None:
            List.append(current.data)
            prev = current
            current = current.next
            
            
        
        
        return print(List)
    
    
#testing
LL = LinkedList()

#getting data
openfile = open('Task3data.txt', 'r')
data = openfile.readlines()
openfile.close()

for i in data:
    i = i.strip()
    
    i = int(i)
    
    LL.insert(i)
    
    
#testing
LL.to_String()
    
    
    
    
    
    
        
        
        

            
            
        
            
        
        
    
        
        
```
