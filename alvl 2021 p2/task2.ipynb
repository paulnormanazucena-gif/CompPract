```python
#task 2.1

def task2_1(filename):
    openfile = open(filename, 'r')
    data = openfile.readlines()
    
    openfile.close()
    List = []
    
    for i in data:
        i.strip()
        List.append(int(i))
        
        
    return List


result = task2_1('TEN.txt')
print(result)
print(len(result))
        
    
    
```

```python
#task 2.2
def task2_2(list_of_integers):
    arr = list_of_integers
    
    for i in range(1, len(arr)):
        
        
        j = i -1
        key = arr[i]
        
        while arr[j] > key and j > -1:
            arr[j+1] = arr[j]
            j -= 1
            
        arr[j+1] = key
        
        
    return arr


print(task2_2(task2_1('TEN.txt')))
```

```python
#task 2.3
def quicksortmid(arr, i, j):
    low = i
    high = j
    
    mid = (low + high) // 2
    pivot = arr[mid]

    
    
    while low <= high:
        while arr[low] < pivot:
            low += 1
            
        while arr[high] > pivot:
            high -= 1
            
        if low <= high:
            arr[low], arr[high] = arr[high], arr[low]
            low += 1
            high -= 1
            
            
    if low < j:
        quicksortmid(arr, low, j)
        
    if high > i:
        quicksortmid(arr, i, high)
        
    return arr
        
        
        
        
        
        
def task2_3(list_of_integers):
    return quicksortmid(list_of_integers, 0, len(list_of_integers)-1)


print(task2_3(task2_1('TEN.txt')))
    
```

```python
import timeit 

print('INSERTION SORT')
#insertion sort
t1000 = task2_1('THOUSAND.txt')
time1000 = timeit.timeit(lambda: task2_2(t1000), number=1)
print('THOUSAND:',time1000)

t100 = task2_1('HUNDRED.txt')
time1000 = timeit.timeit(lambda: task2_2(t1000), number=1)
print('HUNDRED :',time1000)

t10 = task2_1('TEN.txt')
time1000 = timeit.timeit(lambda: task2_2(t1000), number=1)
print('TEN     :' ,time1000)


print()
print('QUICKSORT')

#quicksort sort
t1000 = task2_1('THOUSAND.txt')
time1000 = timeit.timeit(lambda: task2_3(t1000), number=1)
print('THOUSAND:',time1000)

t100 = task2_1('HUNDRED.txt')
time1000 = timeit.timeit(lambda: task2_3(t1000), number=1)
print('HUNDRED :',time1000)

t10 = task2_1('TEN.txt')
time1000 = timeit.timeit(lambda: task2_3(t1000), number=1)
print('TEN     :' ,time1000)



#the quicksort consistenly completes the task faster than the insertion sort
```
