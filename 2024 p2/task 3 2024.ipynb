```python
#task 3.1

def header(sentence):
    string = '<h1>'
    string = string + sentence
    string = string + '</h1>'
    
    return string


def subheader(sentence):
    string = '<h2>'
    string = string + sentence
    string = string + '</h2>'
    
    return string


#def hyperlink(sentence):
#    string = '<a href = "https://www.seab.gov.sg">'
#    string = string + sentence
#    string = string + '</a>'
    
    
def paragraph(sentence):
    string = '<p>'
    string = string + sentence
    string = string + '</p>'
    return string
    


print(header('My element'))
print(subheader('My element'))
print(paragraph('My element'))
    
```

```python
def hyperlink(link_sentence):
    items = link_sentence.split(',')
    url = items[0]
    sentence = items[1].strip()
    
    string = '<a href = "' +   url + '" >'
 
    string = string + sentence
    string = string + '</a>'
    
    return string

    
print(hyperlink('https://www.cambridgeinternational.org,Click Here'))
```

```python
def body(List):
    string = ''
    
    #try:
    for i in List:
        #print(i)
        #print(i.split(':'))
        
        Type, data = i.split(':',1)
        
        data = data.strip()


        if Type == 'heading':
            tobeadd = header(data)
            string = string + tobeadd + '\n'




        elif Type == 'subheading':
            tobeadd = subheader(data)
            string = string + tobeadd + '\n'

        elif Type == 'paragraph': 
            tobeadd = paragraph(data)
            string = string + tobeadd + '\n'

        elif Type == 'hyperlink':  
            tobeadd = hyperlink(data)
            string = string + tobeadd + '\n'


        else:
            return 'function not fond'

    return string
    
    #except:
     #   return 'Try failure'
    
    

        
    
            
            
    
```

```python
#task 3.4

def read_data(filename):
    openfile = open(filename, 'r')
    data = openfile.readlines()
    openfile.close()
    
    List = []
    for i in data:
        
        i = i.strip()
        List.append(i)
        
        
    result = body(List)
        
        
    return result


print(read_data('firstPage.txt'))
        
        
```

```python
def write_data(filename, data):
    
    string = '<!doctype html> '  '\n'
    
    string = string + data
    
    
    
    
    openfile = open(filename+'.html', 'w')
    openfile.write(string)
    openfile.close()
    
    return
    
    
```
