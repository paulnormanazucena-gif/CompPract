```python
#task 1.1
import random

def generatecards():
    cards = []
    for i in range(10):
        number = random.randint(1, 4)
        colour = random.randint(0,2)
        
        colours = ['Red', 'Green', 'Blue']
        colour = colours[colour]
            
            
        card = [number, colour]
        cards.append(card)
        
        
        
    return cards



        
```

```python

def comparehands(hand1, hand2):
    
    score1 = 0
    score2 = 0
    
    for i in range(10):
        card1 = hand1[i]
        card2 = hand2[i]
        
        
        
        
        whowin = compare(card1[1], card2[1])
        
        if whowin == 1:
            score1 = score1 + card1[0] * 2
            
            
        elif whowin == 2:
            score2 = score2 + card2[0] * 2
            
            
        elif whowin == 3:
            score1 = score1 + card1[0]
            
            score2 = score2 + card2[0]
            
    score =[]
    score.append(score1)
    score.append(score2)
    return  score


#mainprogram

    
scores = comparehands(sortp1, sortp2)  
print(scores[0])
print(scores[1])
```

```python
def winner(hand1, hand2):
    scores = comparehands(hand1,hand2)
    
    if scores[0] > scores[1]:
        return 'Player 1 win'
    
    elif scores[0] == scores[1]:
        return 'Players have drawn'
    
    elif scores[0] < scores[1]:
        return 'Player 2 win'
    
    
print(winner(sortp1, sortp2))
```
