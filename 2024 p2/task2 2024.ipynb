```python
#task 2.1
class Node:
    def __init__(self):
        self.data_value = -1
        
        self.left_node = -1
        self.right_node = -1
        
    #getset
    def getdata(self):
        return self.data_value
    
    def setdata(self,data):
        self.data_value = data
        return
    
    
    def getleft(self):
        return self.left_node
    
    def setleft(self, left):
        self.left_node = left
        
        
    def getright(self):
        return self.right_node
    
    def setright(self, right):
        self.right_node = right
        
        
    
    
    
```

```python
#task 2.2

def calculate_nodes(level):
    total = 0
    for i in range(level):
        number =  2 ** i
        
        total = total + number
        string = 'A ' + str(level) + ' level number tree has ' + str(total) + ' nodes'
        
    return print(string)

calculate_nodes(1)
calculate_nodes(3)
calculate_nodes(5)

calculate_nodes(10)

    
```

```python
#task 2.4

def create_empty_tree(listofnodes, numberoflevels):
    total = 0
    for i in range(numberoflevels):
        number =  2 ** i
        
        total = total + number
        
    print(total) 
    
    listofnodes = []
    
    for i in range(total):
        newnode = Node()
        listofnodes.append(newnode)
        
        
    return listofnodes


def numberofleafnodes(levels):
        
    return 1 * 2 ** (levels-1)




        
```

```python
#task 2.5
def assign_leaf_nodes(listofnodes, numberoflevels):
    
    amtleaf = numberofleafnodes(numberoflevels)
    
    for i in range(amtleaf):
        
        listofnodes[i].setdata(i+1)

    return listofnodes

#mainprogram

tree = create_empty_tree([], 3)
result = assign_leaf_nodes(tree, 3)
print(tree)


display = ''
    
for i in result:
    string = str(i.getdata()) + ' ' + str(i.getleft()) + ' ' + str(i.getright())
    display = display + string + '\n'
    
print(display)
    
```

```python
#task 2.6

def create_number_tree(listofnodes, numberoflevels):
    if numberoflevels == 1:

        string = 'left pointer | data | right pointer' + '\n' +  '-1           | 1    | -1 '
        return print(string)
    
    
    listofnodes = create_empty_tree(listofnodes, numberoflevels)
    
    listofnodes = assign_leaf_nodes(listofnodes, numberoflevels)
    
    amtofleaf = numberofleafnodes(numberoflevels)
    
    count = 0
    for i in range(amtofleaf, len(listofnodes)):
    
        left = count
        right = count + 1
        count += 2
        
        #assigns a left and right to a node then makes the data the sum of the data of left and right
        
        
        curnode = listofnodes[i]
        
        
        
        curnode.setleft(left)

     
        
        curnode.setright(right)
        
        
        number = listofnodes[left].getdata() + listofnodes[right].getdata()
        curnode.setdata(number)
    
    display = 'left pointer | data | right pointer' + '\n'
        
    for i in listofnodes:
        string =  str(i.getleft()) + ' ' + str(i.getdata()) + ' ' + str(i.getright())
        display = display + string + '\n'
        
    return print(display)


create_number_tree([], 1)
print()

create_number_tree([], 3)
print()
create_number_tree([], 5)


    
        
```
