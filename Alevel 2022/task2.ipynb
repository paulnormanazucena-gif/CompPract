```python
import random
#task2.1


def task2_1():
    List = []
    for i in range(100):
        
        num = random.randint(0, 100)
        List.append(num)
        
        
    return List
```

```python
#task 2.2

def task2_2(arr):
    
    while True:
        swap =False
        
        for i in range(0, len(arr)-1):
            if arr[i] > arr[i+1]:
                arr[i], arr[i+1] = arr[i+1], arr[i]
                swap = True
                
                
        if swap == False:
            break
            
            
    return arr

randomnums = task2_1()
print
sortednums = task2_2(randomnums)
print(sortednums)
            
```

```python
#task 2.3

def mergesort(arr):
    n = len(arr)
    
    if n <= 1:
        return arr
    
    else:
        mid = n // 2
        
        a = mergesort(arr[mid:])
        b = mergesort(arr[:mid])
        
        return merge(a,b)
    
    
def merge(a,b):
    result = []
    
    i,j = 0, 0
    
    while len(a) > i and len(b) > j:
        if a[i] > b[j]:
            result.append(b[j])
            j+=1
        else:
            result.append(a[i])
            i += 1
            
    while len(a) > i:
        result.append(a[i])
        i += 1
        
        
    while len(b) > j:
        result.append(b[j])
        j+=1
        
        
    return result




#main program
while True:
    
    print("""
    Which sort would you like to use?
    option 1: Bubble Sort
    option 2: Merge Sort
    """)
    option = input('1 or 2?: ')
    
    
    if option != '1' and option != '2':
        pass
    else:
        
        randomnum = task2_1()
        
        if option == '1':
       
            sort = task2_2(randomnum)
            
            print(sort)
            break
            
        else:
        
            sort = mergesort(randomnum)
            print(sort)
            break
            
        





```

```python
#task 2.4

def binarysearch(arr, low, high, key):
    if low < high:
        mid = (low+high) // 2
        
        if arr[mid] == key:
            return mid
        
        elif arr[mid] > key: #since ascending,key is at the left
            
            return binarysearch(arr, low, mid-1, key)
        else:
            return binarysearch(arr, mid+1, high, key)
        
        
    return  -1



#main program

while True:

    option = input('Key: ')
    option = int(option)

    result = binarysearch(sortednums, 0, len(sortednums)-1, option)
    if result == -1:
        result = 'Not found'

    print(result)


```
