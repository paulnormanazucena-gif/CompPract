```python
import sqlite3

connection = sqlite3.connect('LIBRARY.db')

connection.execute("DROP TABLE IF EXISTS Loan")
connection.execute("DROP TABLE IF EXISTS Member")
connection.execute("DROP TABLE IF EXISTS Book")


connection.execute("""
CREATE TABLE Book (
	BookID	INTEGER,
	Title	TEXT,
    Genre	TEXT,
    PRIMARY KEY(BookID)
);""")

connection.execute("""
CREATE TABLE Member (
	MemberNumber	INTEGER,
	FamilyName	TEXT,
	GivenName	TEXT,
	PRIMARY KEY(MemberNumber)
);""")

connection.execute("""
CREATE TABLE Loan (
	LoanID	INTEGER,
	MemberNumber	INTEGER,
	BookID	INTEGER,
	DateLoaned	TEXT,
	Returned	INTEGER,
	FOREIGN KEY(MemberNumber) REFERENCES Member(MemberNumber),
	PRIMARY KEY(LoanID)
);""")


connection.commit()
connection.close()

```

```python
#task 4.2 inserting data

#for book

openfile = open('BOOK.txt', 'r')
Books = openfile.readlines()


openfile.close()


openfile = open('MEMBER.txt', 'r')
Members = openfile.readlines()


openfile.close()



openfile = open('LOAN.txt', 'r')
Loans = openfile.readlines()


openfile.close()

# print(Books)
# print()
# print(Members)
# print()
# print(Loans)

#inserting

connection = sqlite3.connect('LIBRARY.db')

for i in Books:
    
    bookid, name, genre = i.split(',')
    
    connection.execute('INSERT INTO Book(BookID, Title, Genre) VALUES (?,?,?)', (int(bookid), name, genre.strip()))
    
for i in Members:
    memnum, famname, givenname = i.split(',')
    connection.execute('INSERT INTO Member(MemberNumber, FamilyName, GivenName) VALUES (?,?,?)', (int(memnum), famname, givenname.strip()))
    
for i in Loans:
    ID, memnum, bookid, date, returned = i.split(',')
    if returned.strip() == 'FALSE':
        returned = 0
        
    else:
        returned = 1
        
        
    connection.execute('INSERT INTO Loan(LoanID, MemberNumber, BookID, DateLoaned, Returned) VALUES (?,?,?,?,?)', (ID,memnum,bookid, date,returned))
    

connection.commit()
connection.close()
```
