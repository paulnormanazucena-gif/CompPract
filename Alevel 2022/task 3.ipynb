```python
import random
class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None
        
        
    #getset
    def getleft(self):
        return self.left
    def setleft(self,left):
        self.left = left
    def getright(self):
        return self.right
    def setright(self, right):
        self.right = right
    def getdata(self):
        return self.data
    def setdata(self,data):
        self.data  = data
        
        
class Tree:
    def __init__(self):
        self.root = None
        
        
    def isempty(self):
        return self.root == None
    
    
    
    def insert(self, integer):
        newnode = Node(integer)
        
        if self.isempty():
            self.root = newnode
            return
        
        
        currnode = self.root
        prev = None
        
        while currnode != None:
            prev = currnode
            
            
            if currnode.data > integer:
                currnode = currnode.left
                
            else:
                currnode = currnode.right
                
                
        if prev.data > integer:
            prev.left = newnode
            
        else:
            prev.right = newnode
            
    def inorder(self):
        return self.inorderhelper(self.root)
    
    def inorderhelper(self, root):
        if root == None:
            return
        
       
        self.inorderhelper(root.left)
        print(root.data)
        self.inorderhelper(root.right)

    def postorder(self):
        return self.postorderhelper(self.root)
    
    def postorderhelper(self, root):
        if root == None:
            return
        #always rootleft, rootright, only thing thta changes is where to put the print data
        
        self.postorderhelper(root.left)
        self.postorderhelper(root.right)
        print(root.data)
        
#main program
 
BST = Tree()

for i in range(10):
    number = random.randint(0,999)
    BST.insert(number)
print('Inorder')
print('-------')
BST.inorder()
print()

print('Postorder')
print('--------')
BST.postorder()
            
             
        
```
