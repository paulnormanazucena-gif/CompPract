```python
#task1.1

def hashing(character):
    
    
    
    if character == ' ':
        return ord('!')
    
    if character.isalpha() == False:
        return -1
    
    if character == character.lower():
        asciicode = ord(character)
        asciicode = asciicode + 10
        
        if asciicode > 122:
            newcode = 97 + (asciicode-123)
            
            
        else:
            newcode = asciicode
            
            
        return newcode
    
    elif character == character.upper():
        asciicode = ord(character)
        asciicode += 10
        
        if asciicode > 90:
            newcode = 65 + (asciicode - 91)
        else:
            newcode = asciicode
            
            
        return newcode
    
    
 
    
    

            
        
```

```python
#task 1.2


print(hashing('A'))
print(hashing('a'))
print(hashing('#'))
print(hashing(' '))
```

```python
#task 1.3

openfile = open('DATATOENCRYPT.txt', 'r')
data = openfile.readlines()
openfile.close()


print(data)
encrypted = ''

for i in data[0]:
   
    result = hashing(i)
    #print(result)
    
    if result == -1:
        pass
    else:
        encrypted = encrypted + chr(result)

#inserting in new file

openfile = open('ENCRYPTEDMESSAGE.txt', 'w')
openfile.write(encrypted)
openfile.close()
    
```
