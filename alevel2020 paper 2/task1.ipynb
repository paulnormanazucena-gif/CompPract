```python
#task 1,1

def task1_1(string_value):
    h = 0
    for i in range(0, len(string_value)):
                   val = 33  * ord(string_value[i])
                   h = (h  + val) % 1024
                   
    return h
                   
                   
x = task1_1('Hello')
y = task1_1('Hallo')                   
z = task1_1('Hullo')                   
print(x, y, z)
                
```

```python
#task 1.2

def task1_2(seed, string_value):
    string = seed + string_value
    
    h = task1_1(string)
    
    return h

x = task1_2('seed-one','Hello')
y = task1_2('seed-two','Hallo')                   
z = task1_2('seed-three','Hullo')                   
print(x, y, z)

```
