```python
from random import * #From _ import _
#task 2.1

def task2_1(filename, quantity, maximum):
    string = ''
    
    for i in range(quantity):
        
        num = randint(0,maximum) #randint inclusive
        row = str(num) + '\n'
        string = string + row
        
        
    openfile = open(filename, 'w')
    
    openfile.write(string)
    
    openfile.close()
    return

task2_1('randomnumbers.txt', 1000, 5000)

    
```

```python
#task 2.2

def mergesort(arr):
    n = len(arr)
    
    if n == 1:
        return arr
    
    
    
    



def task2_2(list_of_integers):
    n = len(list_of_integers)
    
    if n == 1:
        return list_of_integers
    
    
    else:
        mid = n // 2
        a = task2_2(list_of_integers[:mid])
        b = task2_2(list_of_integers[mid:])
        
        return merge(a,b)
    
    
def merge(a,b):
    result = [None] * (len(a) + len(b))
    
    i,j,k = 0,0,0
    
    while i < len(a) and j < len(b):
        if a[i] > b[j]:
            result[k] = b[j]
            j += 1
            
        else:
            result[k] = a[i]
            i += 1
            
        k += 1
        
        
    while i < len(a):
        result[k] = a[i]
        i += 1
        k += 1
    while j < len(b):
        result[k] = b[j]
        j += 1
        k += 1
        
    return result



test_list = [56,25,4,98,0,18,4,5,7,0]
print(task2_2(test_list))

```

```python
#task 2.3

def task2_3(filename_in, filename_out):
    openfile = open(filename_in, 'r')
    
    data = openfile.readlines()
    
    openfile.close()
    
    list_of_integers = []
    for i in data:
        i = i.strip()
        i = int(i)
        list_of_integers.append(i)
        
    sortedlist = task2_2(list_of_integers)
    
    
    string = ''
    
    for i in sortedlist:
        row = str(i) + '\n'
        string = string + row
        
    
    openfile = open(filename_out, 'w')
    openfile.write(string)
    openfile.close()
    
    
    
    return




#main program
task2_3('randomnumbers.txt', 'sortednumbers.txt')

    
    
        
```
