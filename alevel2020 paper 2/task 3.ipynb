```python
#task 3.1

def task3_1(quantity_of_data):
    number = ''
    unit = ''
    for i in quantity_of_data:
        if i.isalpha():
            unit = unit + i
        else:
            number = number + i
            
            
    number = int(number)
    
    
    if unit == 'KB':
        return print(number * 10 ** 3)
    
    elif unit == 'MB':
        return print(number * 10 ** 6)
        
    elif unit == 'GB':
        return print(number * 10 ** 9)
        
    elif unit == 'TB':
        return print(number * 10 ** 12)
        
    else:
        return print('invalid data')
    
task3_1('8KB')
    
        
```

```python
def task3_2(quantity_of_data):
    number = ''
    unit = ''
    for i in quantity_of_data:
        if i.isalpha():
            unit = unit + i
        else:
            number = number + i
            
            
    number = int(number)
    
    
    if unit == 'KiB':
        return print(number * 2 ** 10)
    
    elif unit == 'MiB':
        return print(number * 2 ** 20)
        
    elif unit == 'GiB':
        return print(number * 2 ** 30)
        
    elif unit == 'TiB':
        return print(number * 2 ** 40)
        
    else:
        return print('invalid data')
    
task3_2('2MiB')
```

```python
def task3_3(quantity_of_data, target_unit):
    list_of_units = ['KB', 'MB', 'GB', 'TB', 'KiB', 'MiB', 'GiB', 'TiB']
    list_of_values = [10 ** 3, 10 ** 6,  10 ** 9, 10 ** 12, 2 ** 10, 2 ** 20, 2 ** 30, 2 ** 40 ]
    
    
    
    number = ''
    unit = ''
    for i in quantity_of_data:
        if i.isalpha():
            unit = unit + i
        else:
            number = number + i
    

    try:   
        index = list_of_units.index(unit)


        byte = float(number) * list_of_values[index]


        index_denominator = list_of_units.index(target_unit)

        result = byte / list_of_values[index_denominator]
        return print(result)

    except:
        return print('invalid data')
    
    

   
    
task3_3('512MiB', 'GiB')
task3_3('ghgh', 'Kib')
task3_3('0.5GiB', 'MiB')
    
```
