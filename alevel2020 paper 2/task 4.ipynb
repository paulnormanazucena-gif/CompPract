```python
import datetime
class Person:
    def __init__(self, name, DoB):
        self.full_name = name
        self.date_of_birth = DoB #YYYY-MM-DD
        return
    
        
        
        
    def is_adult(self):
        today = datetime.date.today() #datetime
        num_equiv = ''
        for i in today:
            if i.isdigi():
                num_equiv = num_equiv + i
                
            else:
                pass
        
        num_equiv = int(num_equiv)
        
        
        bday = ''
        
        for i in self.date_of_birth:
            if i .isdigi():
                bday = bday + i
            else:
                pass
            
            
        bday = int(bday)
        
        
        check = num_equiv - bday
        
        
        if check >= 180000:
            return True
        else:
            return False
            
        
        
    def screen_name(self):
        name = ''
        
        for i in self.full_name:
            if i.isalpha():
                name = name + i
            else:
                pass
            
            
        bday = self.date_of_birth
        month = bday[5:7]
        day = bday[8:]
        
        screenname = name + month + day
        
        
        return screenname
    
    
P = Person('John Tan', '2000-06-01')

print(P.screen_name())
```
