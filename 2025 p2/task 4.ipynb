```python
# Task 4.1
hash_table = [None] * 100
```

```python
# Task 4.2
def calculate_hash(key):
    index = key % 100
    return index
```

```python
# Task 4.3
def insert(key, data):
    global hash_table
    newrecord = [key, data]

    index = calculate_hash(key)

    for i in range(100):
        index = (index + i) % 100
        if hash_table[index] is None or len(hash_table[index]) == 3:
            hash_table[index] = newrecord
            return True
        else:
            pass
    return False
```

```python
# Task 4.4
def read_data():
    with open('records.txt', 'r') as openfile:
        data = openfile.readlines()

    for i in data:
        #print(i)
        key, record = i.split(' ')
        record = record.strip()
        insert(int(key), record)

read_data()
# print(hash_table)
```

```python
# Task 4.5
def find_record(key):
    global hash_table
    index = calculate_hash(key)

    for i in range(100):
        index = (index + i) % 100
        if hash_table[index][0] == key:
            return hash_table[index][1]
        else:
            pass
    return False

# print(find_record(475))
```

```python
# Task 4.6
read_data()
result = find_record(475)

if result == False:
    print('Record Not Found')
else:
    print('Record data: ' + result)
```
