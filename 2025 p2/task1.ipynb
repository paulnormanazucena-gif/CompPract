```python
# Task 1.1

def push(stack, top, data):
    # top is next space
    if len(stack) == top:
        stack.append(None)
    else:
        pass

    stack[top] = data
    top = top + 1
    return stack, top
```

```python
# Task 1.2

def pop(stack, top):
    if top == 0:
        return False

    top -= 1
    item = stack[top]

    return stack, top, item
```
