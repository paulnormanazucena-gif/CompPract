```python
# Task 3.1
import sqlite3

def delete_tables():
    connection = sqlite3.connect('website.db')
    connection.execute('DROP TABLE IF EXISTS User')
    connection.execute('DROP TABLE IF EXISTS Post')
    connection.execute('DROP TABLE IF EXISTS Comment')
    connection.commit()
    connection.close()
    return

def create_tables():
    connection = sqlite3.connect('website.db')
    connection.execute("""
    CREATE TABLE User (
        Username    TEXT,
        Blocked     INTEGER,
        PRIMARY KEY(Username)
    );""")

    connection.execute("""
    CREATE TABLE Post (
        PostID          TEXT,
        PostText        TEXT,
        DateTimePosted  TEXT,
        Username        TEXT,
        PRIMARY KEY(PostID)
    );""")

    connection.execute("""
    CREATE TABLE Comment (
        CommentID           TEXT,
        PostID              TEXT,
        Username            TEXT,
        TheText             TEXT,
        DateTimeCommented   TEXT,
        PRIMARY KEY(CommentID),
        FOREIGN KEY(Username) REFERENCES User(Username),
        FOREIGN KEY(PostID) REFERENCES Post(PostID)
    );""")

    connection.commit()
    connection.close()
    return

# main program
delete_tables()
create_tables()
```

```python
# Task 3.2
delete_tables()
create_tables()

connection = sqlite3.connect('website.db')

# Insert User
with open('user.txt', 'r') as openfile:
    data = openfile.readlines()

for i in data:
    user, blocked = i.split(',')
    connection.execute(
        "INSERT INTO User(Username, Blocked) VALUES (?, ?)",
        (user, int(blocked.strip()))
    )
connection.commit()

# Insert Post
with open('post.txt', 'r') as openfile:
    data = openfile.readlines()

for i in data:
    id, text, datetime, username = i.split(',')
    connection.execute(
        "INSERT INTO Post(PostID, PostText, DateTimePosted, Username) VALUES (?, ?, ?, ?)",
        (id, text, datetime, username.strip())
    )
connection.commit()

# Insert Comment
with open('comment.txt', 'r') as openfile:
    data = openfile.readlines()

for i in data:
    id, postid, username, text, datetime = i.split(',')
    connection.execute(
        "INSERT INTO Comment(CommentID, PostID, Username, TheText, DateTimeCommented) VALUES (?, ?, ?, ?, ?)",
        (id, postid, username, text, datetime.strip())
    )
connection.commit()

# Optional test queries
# print(connection.execute("SELECT * FROM User").fetchall())
# print(connection.execute("SELECT * FROM Post").fetchall())
# print(connection.execute("SELECT * FROM Comment").fetchall())

connection.close()
```

```python
# Task 3.3
def get_blocked_comments():
    connection = sqlite3.connect('website.db')
    results = connection.execute("""
    SELECT Comment.Username, Comment.TheText
    FROM User, Post, Comment
    WHERE User.Blocked = 1 AND User.Username = Comment.Username
    """).fetchall()

    string = ''
    for username, comment in results:
        string += username + ' commented ' + comment + '\n'

    connection.close()
    return string

print(get_blocked_comments())
```
