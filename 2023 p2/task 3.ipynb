```python
def inputPuzzle():
    #create puzzle
    arr = []
    for i in range(9):
        newrow = [None] * 9
        arr.append(newrow)
    
    
#adding the items
    for i in range(9):
        while True:
            row = input('Input row: ')
            

            List = row.split(' ')
            validRow = []
            
            valid = True
            
            if len(List) != 9:
                valid = False
                
                
            if valid == False:
                print('Input a proper row')
            else:
                for number in range(9):
                    check = List[number]
                    #print(check)
                    if int(check) < 10:
                        pass
                    else:
                        valid = False

                    validRow.append(check)



                if valid == False:
                    print('Input a proper row')

                else:



                    for position in range(9):
                        arr[i][position] = int(validRow[position])

                    break
                    
                    
            


    return arr

```
