```python
#task 2.1
class QueueItem:
    def __init__(self, ID, Quantity):
        self.id = ID
        self.qty = Quantity
        
        
    def getID(self):
        return self.id
    
    def getQuantity(self):
        return self.qty
    
    
```

```python
#task2.2
class Queue:
    def __init__(self):
        self.queueList = [None] * 4
        #when empty head and tail are the same
        self.headPointer = 0
        self.tailPointer = 0
        
        self.numberItems = 0 #number of items in list
        
        
    def is_empty(self):
        return self.numberItems == 0
    
    def is_full(self):
        return self.numberItems == 4
    
        
        
    def enqueue(self, QueueItem):
        if self.is_full():
            return False
        
        
        #insert from front, dequeu from back, when inserting, increment the back
        self.queueList[self.tailPointer] = QueueItem
        self.tailPointer = (self.tailPointer + 1) % 4 #CHECK FOR TYPOS WHEN GOT PROBLEM
        self.numberItems += 1
        return True
    
    
    
            
    def dequeue(self):
        if self.is_empty():
            return False
        
        item = self.queueList[self.headPointer]
        
        
        self.headPointer = (self.headPointer + 1 ) % 4
        
        self.numberItems -= 1
        return item
        
        
    
    
    def printQueue(self):
        for i in range(4):
            qi = self.queueList[(self.headPointer+i)%4]
            if qi == None:
                pass
            else:
                ID = qi.getID()
                Qty = qi.getQuantity()
                
                print(ID + ' | ' + Qty)
                
                
        pass
        
```

```python
#task 2.3

Q = Queue()

for i in range(5):
    qi = QueueItem(str(i), str(i+10))
    
    print(Q.enqueue(qi))
```

```python
#task 2.4


for i in range(5):
    print(Q.dequeue())
 
```
