```python
import sqlite3


def createtables():
    connection = sqlite3.connect('TESTDATABASE.db')
    #create student tables
    connection.execute("""
    CREATE TABLE Student (
    StudentID	INTEGER,
    GivenName	TEXT,
    FamilyName	TEXT,
    ClassName	TEXT,
    PRIMARY KEY(StudentID)
    );
    """)
    
    #create test
    connection.execute("""
    CREATE TABLE Test (
    TestID	TEXT,
    MaxMarks	INTEGER,
    PRIMARY KEY(TestID)
    );
    """)
    
    #create class
    connection.execute("""
    CREATE TABLE Class (
    ClassName	TEXT,
    YearGroup	INTEGER,
    PRIMARY KEY('ClassName')
    );
    """)
    
    
    #create student test
    connection.execute("""
    CREATE TABLE Student_Test (
    StudentTestID	INTEGER,
    StudentID	INTEGER,
    TestID	TEXT,
    Mark	INTEGER,
    DateTest	TEXT,
    PRIMARY KEY(StudentTestID),
    FOREIGN KEY(TestID) REFERENCES Test
    );
    """)
    
    connection.commit()
    connection.close()
    
def deletetables():
    connection = sqlite3.connect('TESTDATABASE.db')
    connection.execute('DROP TABLE IF EXISTS Student')
    connection.execute('DROP TABLE IF EXISTS Test')
    connection.execute('DROP TABLE IF EXISTS Class')
    connection.execute('DROP TABLE IF EXISTS Student_Test')
    
    
    
    
    connection.commit()
    connection.close()
    
    
#main program
deletetables()
createtables()
```
