```python
#task 4.1

def readfile(filename):
    openfile = open(filename, 'r')
    data = openfile.readlines()
    openfile.close()
    values = []
    for i in data:
        i = i.strip()
        values.append(i)
        
        
    return values



def compress(values):
    compressedvalues = ''
    currentitem = None
    count = 0
    for i in values:
        
        if count == 9:
            string = currentitem + str(count)
            compressedvalues = compressedvalues + string
            
            currentitem = i
            count = 1
            
        
        
        elif currentitem == None:
            currentitem = i
            count = 1
            
        elif currentitem == i:
            count +=1
            
        else:
            string = currentitem + str(count)
            compressedvalues = compressedvalues + string
            
            currentitem = i
            count = 1
    string = currentitem + str(count)
            
            
    compressedvalues = compressedvalues + string
    
    return compressedvalues
            
        
    
    
    
def write(compressedvalues):
    openfile = open('compressedimage.txt', 'w')
    openfile.write(compressedvalues)
    openfile.close()
    return

#<td style="background-color: {{ color }};"></td>

def decompress(compressedvalues):
    decompressed = ''
    for i in range(0,len(compressedvalues),4):
        first, second, third ,fourth = i ,i + 1, i + 2, i + 3
        binary = compressedvalues[first] +compressedvalues[second] + compressedvalues[third]
        #print(binary)
        
        decomp = binary * int(compressedvalues[fourth])
        decompressed = decompressed + decomp
    return decompressed


def writedec(decompressed):
    openfile = open('deccompressedimage.txt', 'w')
    openfile.write(decompressed)
    openfile.close()
    return
    

#main program

values = readfile('imagefile.txt')
compressedvalues = compress(values)
write(compressedvalues)
#print('worked')
print(len(values))

decompressedvalues = decompress(compressedvalues)
writedec(decompressedvalues)
#print('hi')
```
