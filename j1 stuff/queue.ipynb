```python
#enqueue
#dequeue
#is_empty
#is_full
#peek
#size
#display

#enqueue rear is updated
#dequeue front is update
class queue:
    #constructor
    def __init__(self,size):
        self.size = size
        self.arr = [None] * self.size
        self.front = -1
        self.rear = -1
        
        
    def is_empty(self):
        if self.front == -1 and self.rear == -1: #self.front = self.rear dont do because i will just reset it when dequeueing
            return True
        
        else:
            return False
        
        
    def is_full(self):
        if self.rear == self.size -1:
            return True
        
        else:
            return False
        
        
    def enqueue(self, item): #move rear forward
        if self.is_full():
            return False
        
        if self.is_empty(): #need to update first
            self.front += 1
            self.rear += 1
            
            self.arr[self.rear] = item
            
            
        else:
            
            self.rear += 1
            
            self.arr[self.rear] = item
            return True
            
    def dequeue(self): #move front forward
        
        if self.is_empty():
            return False
        
        else:
            data = self.arr[self.front]
            self.front +=1
            
            if self.front == self.rear +1:
                self.front = -1
                self.rear = -1
                
            else:
                pass
            
            return data
        
    def display(self):
        print('Data,    test')
        
        for i in range(len(self.arr)):
            if i == self.front:
                print(self.arr[i], ' FRONT')
                
            elif i == self.rear:
                print(self.arr[i], ' REAR')
                
            else:
                print(self.arr[i])
                
                
    
#test
# Create a queue of size 5
q = queue(5)

print("Initially empty?", q.is_empty())   # Expect True
print("Initially full?", q.is_full())     # Expect False

# Enqueue some items
print("\nEnqueueing items...")
q.enqueue("A")
q.enqueue("B")
q.enqueue("C")
q.display()

# Peek at front
print("\nFront element should be A:", q.arr[q.front])

# Dequeue one item
print("\nDequeued:", q.dequeue())         # Expect "A"
q.display()

# Enqueue more items
print("\nEnqueueing more...")
q.enqueue("D")
q.enqueue("E")
q.display()

# Check if full
print("\nIs full?", q.is_full())          # Expect True

# Try enqueue when full
print("\nTrying to enqueue F (should fail):", q.enqueue("F"))

# Dequeue all items
print("\nDequeuing all...")
while not q.is_empty():
    print("Dequeued:", q.dequeue())
    q.display()

# Final check
print("\nEmpty after all dequeues?", q.is_empty())  # Expect True
    
    

        
        
    
```
