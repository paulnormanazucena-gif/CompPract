```python

balance = 0


while True:
    
    print()
    print("""
    1: view balance
    2: deposit
    3: withdraw
    4: quit""")
  
    
    option = input('Which action will you take?(1-4): ')
    print()
    print()
    
    if option == '4':
        break
    
    elif option == '3':
        amount = input('How much?: ')
        if float(amount) > balance:
            print('Invalid request')
        else:
            balance = balance - float(amount)
            print('Cash has been withdrawn, remaining balance = $' + str(balance))
            
        
    elif option == '2':
        amount = input("How much?: ")
        balance = balance + float(amount)
        print("Deposit complete, balance = $" + str(balance))
        
        
        
        
        
    elif option == '1':
        print("Your current balance is $" + str(balance))
        
```
