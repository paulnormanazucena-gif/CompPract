```python
def quickmiddle(arr, i, j):
    low = i
    high = j
    
    #pivot is the middle
    mid = (high + low) // 2
    pivot = arr[mid]
    
    
    while low <= high: #base case
        
        #scan from left
        
        while arr[low] < pivot:
            low += 1
            
            
        while arr[high] > pivot:
            high -= 1
            
            
        if low <= high:
            #swap low and high at the end
            #then increment low and high again
            
            
            arr[high], arr[low] = arr[low], arr[high]
            
            low += 1
            
            high -= 1
            
            
    #recursively sort both halves
    
    
    if i < high:
        quickmiddle(arr, i, high)
        
        
    if j > low:
        quickmiddle(arr, low, j)
        
        
    return arr

numbers = [
  42, 7, 88, 15, 63, 29, 94, 36, 2, 75,
  18, 53, 81, 24, 60, 11, 97, 33, 5, 70,
  27, 49, 91, 14, 38, 66, 22, 83, 31, 9,
  57, 99, 19, 46, 72, 25, 84, 34, 6, 55,
  93, 16, 40, 68, 21, 85, 32, 4, 50, 96,
  12, 44, 78, 28, 62, 90, 17, 35, 73, 3,
  59, 20, 86, 30, 8, 47, 100, 26, 64, 92,
  13, 39, 76, 1, 58, 95, 23, 41, 80, 10,
  56, 98, 37, 67, 45, 82, 71, 48, 79, 54,
  87, 43, 74, 52, 89, 61, 77, 51, 69, 65
]


sort = quickmiddle(numbers, 0, len(numbers)-1)
print(sort)        
        
        
        #pivot low, high pivot
        
```
