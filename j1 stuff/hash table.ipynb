```python
#hash table linear probing is oopen addressing 

#definition: data structure  that is an unordered collection of key value pairs

#key-value pairs key refers to the index of an array and the value is the element sotred in at index

#definiting data of a hash table: both data itsef and function/methods relating to hashtable

#most important part and why: hashing functions as it enables data to be inserted and searched on the hash table


#Alevel: hash table works/wraps around array using modulus operator




#use mod/remainder function to wrap around     index = key % size



#because of mod operator, you are bound to have collsiisions, need to incremeent by 1 until either find a free space, or loops back around




#if got 10k items, make size more than 10k, minimise collisioin and also prevents overflow
#try use prime number, tends to have uniform or random distribution of keys
```

```python
class hashtable:
    def __init__(self, size):
        self.arr = [None] * size
        self.size = size
        
        
        
    def hashfunc(self, key):
        index = key % self.size
        
        
        return index
    
    
    
    def insert(self, kvpair):
        key = kvpair[0]
        
        index = self.hashfunc(key)
        
        
        for i in range(0, self.size):
          
            curindex = (index + i) % self.size
            currentspace = self.arr[curindex]


            if currentspace == None or currentspace == 'deleted':
                self.arr[curindex] = kvpair
                return 'inserted'
                
                
            else:
                pass
            
            
        return 'Failed to insert'
    
    
    def delete(self, kvpair):
        key = kvpair[0]
        index = self.hashfunc(key)
        
        
        for i in range(0, self.size):
            
            curindex = (index + i) % self.size
            
            
            
            currentspace = self.arr[curindex]
            
            if currentspace == kvpair:
                self.arr[curindex] = 'deleted'
                return 'deleted'
                
            else:
                pass
            
            
            
            
        return 'could not be found'
                
                
            
            



```
