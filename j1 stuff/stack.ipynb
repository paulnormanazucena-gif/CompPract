```python
#push
#pop
#peek
#is_empty
#count()


class stack:
    def __init__(self, size):
        self.capacity = size
        self.front = -1
        self.arr = [None] * self.capacity
        
    def is_empty(self):
        return self.front == -1
    
    def is_full(self):
        return self.front == len(self.arr) -1
    
    
    def push(self, item):
        if self.is_full():
            return False
        
        else:
            self.front += 1
            self.arr[self.front] = item
            
            
            return True
        
        
    def pop(self):
        if self.is_empty():
            return False
        
        
        else:
            data = self.arr[self.front]
            self.front -= 1
            
            return data
        
        
        
    def peek(self):
        
        if self.is_empty():
            return 'Empty'
        
        else:
            return self.arr[self.front] 
    
    
    def size(self):
        return self.front + 1
    
    
    
    

            
```
