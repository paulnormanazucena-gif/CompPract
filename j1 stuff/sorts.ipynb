```python
#bubblesort, while then for loop

def bubblesort(arr):
    
    
    while True: #repeats until no more swaps hae been done
        is_sorted = True
        
        for i in range(0, len(arr)-1):
            if arr[i] > arr[i+1]:
                arr[i], arr[i+1] = arr[i+1], arr[i]
                is_sorted = False
                
            else:
                pass
            
            
        if is_sorted == True:
            break
        else:
            pass
    return arr


bubble = bubblesort(numbers)
print(bubble)
        
```

```python
#insertion sort, for then while loop


def insertionsort(arr):
    for i in range(1, len(arr)):
        key = arr[i] #current item
        j = i-1
        
        while j > -1 and arr[j] > key: #while j greater than 1 and the left item is greated than current item
            arr[j+1] = arr[j]
            j -= 1
        
        #place ket at its correct place
        arr[j+1] = key
        
        
    return arr


insert = insertionsort(numbers)
print(insert)
        
    
```

```python
#mergesort

def mergesort(arr):
    n = len(arr)
    
    if n == 1:
        return arr
    else:
        mid = n // 2
        
        a = mergesort(arr[mid:])
        b = mergesort(arr[:mid])
        
        return merge(a,b)
    
def merge(a,b):
    size = len(a) + len(b)
    
    results = [None] * size
    
    i, j, k = 0,0,0
    
    while i < len(a) and j < len(b):
        if a[i] > b[j]:
            results[k] = b[j]
            j += 1
            
        else:
            results[k] = a[i]
            i += 1
            
        k += 1
        
    while i <len(a):
        results[k] = a[i]
        i += 1
        k += 1
        
    while j < len(b):
        results[k] = b[j]
        j += 1
        k += 1
        
        
    return results


merge = mergesort(numbers)
print(merge)
    
```

```python
def quicksort_last(arr,low,high):
    if low < high:
        pivot = partition(arr, low, high)
        quicksort_last(arr, low, pivot-1)
        quicksort_last(arr, pivot+1, high)
        
    return arr

def partition(arr,low,high):
    pivot = arr[high]
    left = low-1
    
    for i in range(low, high):
        if arr[i] <= pivot:
            left =left +1
            arr[i], arr[left] = arr[left], arr[i] #swap arr[i] and arr[left]
            
    arr[left +1], arr[high] = arr[high], arr[left+1]
        
    return left +1


quicklast = quicksort_last(numbers, 0, len(numbers)-1)
print(quicklast)
            
            
```

```python
def quicksortmiddle(arr, i, j):
    low = i
    high = j
    pivot = arr[ (i+j) //2] # integer midpoint
    
    while low <= high:
        #san from left
        
        while arr[low] < pivot:
            low += 1
            
            
        #scan from right
        
        while arr[high] > pivot:
            high -= 1
            
        # swwap if necessary
        
        if low <= high:
            arr[low], arr[high] = arr[high], arr[low]
            
            low += 1
            
            high -= 1
            
    # revursively sort both halves
    
    
    if i < high:
        quicksortmiddle(arr,i, high)
        
    if low < j:
        quicksortmiddle(arr, low,j)
        
        
    return arr
quickmid = quicksortmiddle(numbers, 0, len(numbers)-1)
print(quickmid)
```
