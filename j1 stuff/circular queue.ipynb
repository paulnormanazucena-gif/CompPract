```python
#enqueue
#dequeue
#is_empty
#is_full
#peek
#size
#display

#enqueue rear is updated
#dequeue front is update
class circular_queue:
    #constructor
    def __init__(self,size):
        self.size = size
        self.arr = [None] * self.size
        self.front = -1
        self.rear = -1
        
        
    def is_empty(self):
        if self.front == -1 and self.rear == -1: #self.front = self.rear dont do because i will just reset it when dequeueing
            return True
        
        else:
            return False
        
        
    def is_full(self):
        if (self.rear + 1) % self.size ==  self.front:
            return True
        
        else:
            return False
        
        
    def enqueue(self, item): #move rear forward
        if self.is_full():
            return False
        
        if self.is_empty(): #need to update first
            self.front += 1
            self.rear += 1
            
            self.arr[self.rear] = item
            
            
        else:
            
            self.rear = (self.rear+1) % self.size
            
            self.arr[self.rear] = item
            return True
            
    def dequeue(self): #move front forward
        
        if self.is_empty():
            return False
        
        else:
            
            if self.front == self.rear: #1 left
                
                data = self.arr[self.front]
                self.front = -1
                self.rear =-1 
            else:
                data = self.arr[self.front]
                
                self.front = (self.front +1) % self.size
            
           
            
            return data
        
    def display(self):
        print('Data,    test')
        
        for i in range(len(self.arr)):
            if i == self.front:
                print(self.arr[i], ' FRONT')
                
            elif i == self.rear:
                print(self.arr[i], ' REAR')
                
            else:
                print(self.arr[i])
                
    
```
