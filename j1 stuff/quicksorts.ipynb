```python
numbers = [
  42, 7, 88, 15, 63, 29, 94, 36, 2, 75,
  18, 53, 81, 24, 60, 11, 97, 33, 5, 70,
  27, 49, 91, 14, 38, 66, 22, 83, 31, 9,
  57, 99, 19, 46, 72, 25, 84, 34, 6, 55,
  93, 16, 40, 68, 21, 85, 32, 4, 50, 96,
  12, 44, 78, 28, 62, 90, 17, 35, 73, 3,
  59, 20, 86, 30, 8, 47, 100, 26, 64, 92,
  13, 39, 76, 1, 58, 95, 23, 41, 80, 10,
  56, 98, 37, 67, 45, 82, 71, 48, 79, 54,
  87, 43, 74, 52, 89, 61, 77, 51, 69, 65
]
```

```python
def qm(arr, i, j):
    low = i
    high = j
    
    mid = (low + high) // 2
    
    pivot = arr[mid]
    
    
    while low <= high:
        while arr[low] < pivot:
            low += 1
            
        while arr[high] > pivot:
            high -= 1
            
            
        if low <= high:
            arr[low], arr[high] = arr[high], arr[low]
            
            low += 1
            high -= 1
            
            
    if low < j:
        qm(arr, low, j)
        
    if high > i:
        qm(arr, i, high)
        
        
    return arr


sortqm = qm(numbers, 0, len(numbers)-1)


print(sortqm)
```

```python
def ql(arr, low, high):
    if low < high:
        pivot = p(arr, low, high)
        ql(arr, low, pivot-1)
        ql(arr, pivot+1, high)
        
    return arr


def p(arr, low, high):
    pivot = arr[high]
    left = low-1
    
    
    for i in range(low, high):
        if arr[i] < pivot:  #always check current item to pivot
            left += 1
            #increment first
            #swap current and left
            
        
            arr[left], arr[i] = arr[i], arr[left]
            
        
        
        
    arr[ left+1], arr[high] = arr[high], arr[left+1]
    return left + 1


sortql = ql(numbers, 0, len(numbers)-1)
print(sortql)
            
```

```python
def qf(arr, low, high):
    if low <high:
        pivot = pf(arr, low, high)
        qf(arr, pivot+ 1, high)
        qf(arr, low, pivot -1 )
        
    return arr
        
        
def pf(arr, low, high):
    pivot = arr[low]
    right = high + 1
    
    for i in range(low, high):
        if arr[i] > pivot:
            #for last pivot no need to increment right
            
            arr[i], arr[right] = arr[right], arr[i]
            
            
    arr[right-1], arr[low] = arr[low], arr[right-1]
    
    return right -1


sortqf = qf(numbers, 0, len(numbers)-1)

print(sortqf)
    
        
```
