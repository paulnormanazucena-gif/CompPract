```python
class node:
    #constructor
    def __init__(self, data):
        self.next = -1
        self.data = data
        
    #getset
    
    
class LLstatic:
    #constructor
    def __init__(self, size):
        self.size = size
        
        # setting up the array
        #############################################
        self.arr = [None] * self.size
        
        for i in range(size):
            newnode = node('')
            self.arr[i] = newnode
            
            self.arr[i].next = i + 1
            
        self.arr[-1].next = -1
        
        
        #############################################
        
        self.head = -1
        self.nf = 0
        
        
    def is_empty(self): #check empty using head
        return self.head == -1
    
    
    def is_full(self): #check full using next free
        return self.nf == -1
    
    
    
    
    def insert(self, item):
        
        if self.is_full():
            return False
        
        
        if self.is_empty(): # need make the node to the head
            
            newnode = self.nf #helps to identify
            
            self.nf = self.arr[self.nf].next #increment the nf
            
            self.head = newnode #make newnode the head
            
            self.arr[newnode].data = item # add the item
            
            
            self.arr[newnode].next = -1 #making the head/first item the
            #ONlY ITEM
            
        
            return True
        
        
        else:
            
            current = self.head
            prev = None
            
            while current != -1 and self.arr[current].data < item:
                prev = current
                current = self.arr[current].next
                
            #current is either -1 or the index of item that is bigger than item
            
            
    
                
  
            
            #case 1 insert in head
            
            if prev == None:
                newnode = self.nf #making new node
                self.nf = self.arr[self.nf].next #increment the nf
                
                self.arr[newnode].next = current #connecting them
                
                self.head = newnode #makiing new item the head
                
                self.arr[newnode].data = item #inserts data
                
                return True
            
            #case 2 insert at the end
            elif current == -1: #insert at the end
                newnode = self.nf #making new node
                self.nf = self.arr[self.nf].next #increment nf
                
                self.arr[prev].next = newnode
                
                self.arr[newnode].data = item
                
                self.arr[newnode].next = -1  # nneed to make it the end of the list
                
                
                return True
                
            #insert in themiddle of two nodes
            else:
                newnode = self.nf
                self.nf = self.arr[self.nf].next#increment nf
                
                
                
                self.arr[newnode].data = item #insert the data
                
                #connected the nodess
                self.arr[prev].next = newnode
                self.arr[newnode].next = current
                return True
                
    
    def delete(self, item):
        
        current = self.head
        prev = None
        
        
        while current != -1 and self.arr[current].data != item:
            prev = current
            current = self.arr[current].next
            
            
        #either ends when current == - 1 or the data  is found
        
        
        if current == -1:
            return False
        
        if self.arr[current].data == item:
            
            #case 1 delete the heaad
            if current == self.head:
                self.head = self.arr[current].next #connecting the nodes 
                
                self.arr[current].next = self.nf
                
                self.nf = current
                
                #return thenf back to the nf
                
                
                return True
            
            
            #case 2 delete the end
            

            # case 3 every other case
            else:
                self.arr[prev].next = self.arr[current].next
                
                self.arr[current].next = self.nf
                self.nf = current
                #returning the nf back to the nf
                return True
                
                
                
                
                
                
                
                
                
          
                
            
        
```
