```python
def ql(arr, low, high):
    # low < high 
    
    
    if low < high:
        pivot = partition(arr, low, high)
        ql(arr, low, pivot-1)
        ql(arr, pivot+ 1, high)
        
        
    return arr


def partition(arr, low, high):
    pivot = arr[high] #make pivot the item at the high index
    
    
    left = low - 1 #left then will start scanning from left
    
    
    #scan from left
    
    
    for i in range(low, high):
        if arr[i] <= pivot:  #if current item less than pivot
            
            left = left + 1
            
            arr[i], arr[left] = arr[left], arr[i]
            #swap current item  with arr[left]
            
            
        arr[left+1], arr[high] = arr[high], arr[left+1] #swithc high and left + 1 items
        
    #make sure the return not in tthe for loop   
    return left + 1
    
    
    
numbers = [
  42, 7, 88, 15, 63, 29, 94, 36, 2, 75,
  18, 53, 81, 24, 60, 11, 97, 33, 5, 70,
  27, 49, 91, 14, 38, 66, 22, 83, 31, 9,
  57, 99, 19, 46, 72, 25, 84, 34, 6, 55,
  93, 16, 40, 68, 21, 85, 32, 4, 50, 96,
  12, 44, 78, 28, 62, 90, 17, 35, 73, 3,
  59, 20, 86, 30, 8, 47, 100, 26, 64, 92,
  13, 39, 76, 1, 58, 95, 23, 41, 80, 10,
  56, 98, 37, 67, 45, 82, 71, 48, 79, 54,
  87, 43, 74, 52, 89, 61, 77, 51, 69, 65
]


sort = ql(numbers, 0, len(numbers)-1)
print(sort)


            
```

```python
def ql2(arr, low, high):
    if low < high:
        pivot = p2(arr, low, high)
        ql2(arr, low, pivot -1)

        ql2(arr, pivot+1, high)
        
    return arr


def p2(arr, low, high):
    pivot = arr[high]
    
    left = low - 1
    
    
    for i in range(low, high):
        if arr[i] <= pivot:
            left = left  + 1
            
            arr[left], arr[i] = arr[i], arr[left]
            
     #swwap left plus 1 and high       
    arr[left+1], arr[high] = arr[high], arr[left+1]
    
    
    return left + 1

sort2 = ql2(numbers, 0, len(numbers)-1 ) #need the minus one as 0 base index python
print(sort2)
```

```python
def ql3(arr, low, high):
    if low < high:
        pivot = p3(arr, low, high)
        ql3(arr, low, pivot-1)
        ql3(arr, pivot+1, high)
        
    return arr


def p3(arr, low, high):
    pivot = arr[high]
    left = low - 1  #left = low -1
    
    
    
    for i in range(low, high):
        if arr[i] < pivot:
            left += 1
            arr[i], arr[left] = arr[left], arr[i]
            
    arr[left+1], arr[high] = arr[high], arr[left+1]
    
    return left + 1#return left plus 1
    
    
    
    
    
sort3 = ql3(numbers, 0, len(numbers)-1 ) #need the minus one as 0 base index python
print(sort3)
    
    
    
```
