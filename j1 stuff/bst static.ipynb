```python

 #static means use array, ergo use indexing
class treenode:
    def __init__(self, data):
        self.left = -1
        self.right = -1
        self.data = data


class bst_static:
    def __init__(self, size):
        self.size = size
        
        # setting up the array
        ##################################################################
        self.arr = [None] * size
        #adding the tree nodes
        
        
        for i in range(len(self.arr)):
            self.arr[i] = treenode('')
            
        
        # chain all the tree nodes together
        
        for i in range(len(self.arr)):
            self.arr[i].left = i+1
            
            
        self.arr[-1].left = -1
        #################################################################
        
        #next free
        self.nf = 0
        
        #root
        self.root = -1
        
        
    def is_full(self):
        return self.nf == -1
    
    
    def is_empty(self):
        return self.nf == 0 #
    
    
    
    def insert(self,data):
        
        #case 1: check if full
        
        if self.is_full():
            return False
        
        
        #case 2: if bst is empty, newnode becomes the root node
        
        
        if self.is_empty():
            self.root = self.nf #making the node the root node
            
            self.nf = self.arr[self.nf].left # THIS INCREMENTs THE NEXT FREE
            
            self.arr[self.root].left = -1 # makes the newnode into a normal node
            
            self.arr[self.root].data = data # adds the item
            
            return True
        
        
        
        #case 3, all other cases, means already have some item
        
        newnode = self.nf #makes this easier
        
      
        #WHILE LOOP TO ADD the item
        current = self.root
        prev = None
        movedLeft = False
        
        
        while current != -1: #while not reached the traversed node path
            prev = current
            
            if data > self.arr[current].data: # go to the right
                current = self.arr[current].right
                movedLeft = False
                
            elif data < self.arr[current].data:
                current = self.arr[current].left
                movedLeft = True
                
            else:
                return False
            
        #check to see where to add
        
        if movedLeft == True:
            self.arr[prev].left = newnode #newnode is an index poitning to the node in the array
        else:
            self.arr[prev].right = newnode
            
            
        #doing all the updating to keep it working
        self.arr[self.nf].data = data #adds the item
        
        
        
        self.nf = self.arr[self.nf].left # increments the nf
        
        self.arr[newnode].left = -1 # makes the new node a normal node
        
        
            
        return True
    
    
    
    
    
    
    
            
        
        
        
        
        
        
        
        
        
        
```
