```python
# need tombstone value

class hashtable:
    def __init__(self, size):
        self.arr = [None] * size
        self.size = size
        
        
        
        
        
    def hashfunc(self, kvpair):
        key = kvpair[0]
        index = key % self.size
        return index
        
        
        
    def insert(self, kvpair):
        index = self.hashfunc(kvpair)
        
        for i in range(self.size):
            cur_index = (index + i) % self.size  #need mod cus need loop back
            
            
            if self.arr[cur_index] == None or self.arr[cur_index] == 'deleted': # check if its none or already deleted
                self.arr[cur_index] = kvpair
                return True
            
            else:
                pass
        return False
    
    
    
    def delete(self, kvpair):
        index = self.hashfunc(kvpair)
        
        
        for i in range(self.size):
            cur_index = (index + i) % self.size #need mod cus need loop back

            
            
            if self.arr[cur_index] != None and self.arr[cur_index][0] == kvpair[0]: #check if the  key matches then delete
                self.arr[cur_index] = 'deleted'
                
                return True
            
        return False
    
    
```
