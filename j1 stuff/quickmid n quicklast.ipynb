```python
def quickmid(arr, i, j):
    low = i
    high = j
    
    mid = (low + high) // 2
    
    pivot = arr[mid]
    
    
    while low <= high:
        while arr[low] < pivot:
            low += 1
            
            
        while arr[high] > pivot:
            high -=1
            
            
        if low <= high:
            arr[low], arr[high] = arr[high], arr[low]
            low += 1
            high -= 1
            
    #if low < j
    if low < j:
        quickmid(arr, low, j)
        
    if high > i:
        quickmid(arr, i, high)
        
        
    return arr


sortmid = quickmid(numbers, 0, len(numbers)-1)
print(sortmid)
```

```python
def quicklast(arr, low, high):
    if low < high:
        pivot = partition(arr, low, high)
        quicklast(arr, pivot + 1, high)
        quicklast(arr, low, pivot-1)
        
    return arr



def partition(arr, low, high):
    
    pivot = arr[high]
    left = low -1
    
    
    for i in range(low, high):
        if arr[i] < pivot:
            
            #increment first then swap
            
            
            left += 1
            arr[i], arr[left]= arr[left], arr[i]
            
            
    arr[left + 1], arr[high] = arr[high], arr[left + 1]
    
    
    return left + 1


sortlast = quicklast(numbers, 0, len(numbers)-1)
print(sortlast)
        
```
